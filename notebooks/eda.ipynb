# Exploratory Data Analysis

Run this after downloading data into `data/raw/` (real mode) or directly against the synthetic
generator (default). Produces: tag/genre frequency plots, track-length distribution, and a
chroma/MFCC visualization for one track — a starting point for the report's dataset section.


```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np
import matplotlib.pyplot as plt

from src.synthetic_data import generate_synthetic_dataset, GENRE_NAMES

tracks = generate_synthetic_dataset(n_tracks=240, n_genres=8, seed=42)
print(f'{len(tracks)} tracks generated')
```

## Genre distribution

```python
genre_counts = {}
for t in tracks:
    genre_counts[t.genre] = genre_counts.get(t.genre, 0) + 1

plt.figure(figsize=(7,4))
plt.bar(genre_counts.keys(), genre_counts.values())
plt.title('Genre distribution (synthetic)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Tag frequency (top 20)

```python
from src.synthetic_data import build_top_k_tag_vocab

vocab = build_top_k_tag_vocab(tracks, k=20)
tag_counts = {tag: 0 for tag in vocab}
for t in tracks:
    for tag in t.tags:
        if tag in tag_counts:
            tag_counts[tag] += 1

plt.figure(figsize=(9,4))
plt.bar(tag_counts.keys(), tag_counts.values())
plt.title('Top-20 tag frequency (synthetic)')
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

## Track length (number of segments) distribution

```python
lengths = [len(t.segments) for t in tracks]
plt.figure(figsize=(6,4))
plt.hist(lengths, bins=15)
plt.xlabel('segments per track (5s windows)')
plt.ylabel('count')
plt.title('Track length distribution')
plt.show()
```

## Chroma heatmap for one track

```python
track = tracks[0]
chroma_matrix = np.stack([s.chroma for s in track.segments], axis=1)  # [12, n_segments]

plt.figure(figsize=(8,3))
plt.imshow(chroma_matrix, aspect='auto', origin='lower', cmap='magma')
plt.yticks(range(12), 'C C# D D# E F F# G G# A A# B'.split())
plt.xlabel('segment index')
plt.title(f'Chroma over segments — {track.track_id} ({track.genre})')
plt.colorbar(label='chroma energy')
plt.tight_layout()
plt.show()
```
